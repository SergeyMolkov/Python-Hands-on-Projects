## Task: Take the article (see below) and summarize it using several summarization methods and provide the best summary.
url = https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7

```python
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

medium_url = "https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7"
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get(medium_url)
time.sleep(5)
html = driver.page_source
driver.quit()

soup = BeautifulSoup(html, "html.parser")
```

```python
article = soup.find('article')
if article:
    paragraphs = article.find_all('p')
else:
    paragraphs = soup.find_all('p')

paragraphs = [p for p in paragraphs if len(p.get_text(strip=True).split()) > 3]

for p in paragraphs[:5]:
    print(p.get_text(strip=True))
    print('---')

medium_text = ' '.join(p.get_text(strip=True) for p in paragraphs)
print("Article length:", len(medium_text.split()), "words")
```

```python
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer
from sumy.summarizers.lex_rank import LexRankSummarizer
from sumy.summarizers.luhn import LuhnSummarizer
from sumy.summarizers.lsa import LsaSummarizer


from summa import summarizer

import nltk
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)   # нужен для NLTK 3.8.2+

medium_parser = PlaintextParser.from_string(medium_text, Tokenizer("english"))

summaries = {}
summaries["Sumy - TextRank"] = " ".join(str(s) for s in TextRankSummarizer()(medium_parser.document, 6))
summaries["Sumy - LexRank"]  = " ".join(str(s) for s in LexRankSummarizer()(medium_parser.document, 6))
summaries["Sumy - Luhn"]     = " ".join(str(s) for s in LuhnSummarizer()(medium_parser.document, 6))
summaries["Sumy - LSA"]      = " ".join(str(s) for s in LsaSummarizer()(medium_parser.document, 6))
summaries["Summa"] = summarizer.summarize(medium_text, ratio=0.15)

for name, s in summaries.items():
    print(f"=== {name} ({len(s.split())} words) ===")
    print(s)
    print()
```

```python
best_name = "Sumy - TextRank"
best_summary = summaries[best_name]

with open("SergeyMolkov_summary.txt", "w", encoding="utf-8") as f:
    f.write(f"Best summarizer: {best_name}\n\n")
    f.write(best_summary)

print(summaries[best_name])
```
